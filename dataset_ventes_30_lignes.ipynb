 Identifier la région générant le plus de revenus et les produits qui y contribuent le plus.

```python
# 1. Identifier la région générant le plus de revenus
revenu_par_region = df.groupby("Région")["Revenu_USD"].sum()
meilleure_region = revenu_par_region.idxmax()
revenu_max_region = revenu_par_region.max()

print(f"La région la plus rentable est : {meilleure_region} avec un revenu total de {revenu_max_region:,.2f} USD\n")

# 2. Filtrer le DataFrame sur cette meilleure région
df_top_region = df[df["Région"] == meilleure_region]

# 3. Calculer la contribution de chaque produit dans cette région
produits_top_region = (
    df_top_region.groupby("Produit")["Revenu_USD"]
    .sum()
    .reset_index()
    .sort_values(by="Revenu_USD", ascending=False)
)

# 4. Calculer le pourcentage de contribution par rapport au total de la région
produits_top_region["Contribution (%)"] = (
    produits_top_region["Revenu_USD"] / revenu_max_region * 100
).round(2)

# Affichage des meilleurs produits contributeurs
produits_top_region.head(10)
```

 Calculer l'écart-type du revenu par vente pour évaluer la dispersion des ventes.

```python
# 1. Calcul de l'écart-type global
ecart_type_revenu = df["Revenu_USD"].std()
revenu_moyen = df["Revenu_USD"].mean()
revenu_median = df["Revenu_USD"].median()

# 2. Coefficient de variation (dispersion relative)
coef_variation = (ecart_type_revenu / revenu_moyen) * 100

print(f"Écart-type du revenu : {ecart_type_revenu:,.2f} USD")
print(f"Revenu moyen : {revenu_moyen:,.2f} USD")
print(f"Revenu médian : {revenu_median:,.2f} USD")
print(f"Coefficient de variation : {coef_variation:.2f}%")
```

 Comparer le revenu moyen entre Kinshasa et Lubumbashi et indiquer laquelle des deux
régions performe le mieux.

```python
# 1. Filtrer les données pour Kinshasa et Lubumbashi
df_villes = df[df["Région"].isin(["Kinshasa", "Lubumbashi"])]

# 2. Calculer le revenu moyen par région
revenu_moyen_villes = df_villes.groupby("Région")["Revenu_USD"].mean()

moyenne_kinshasa = revenu_moyen_villes.get("Kinshasa", 0)
moyenne_lubumbashi = revenu_moyen_villes.get("Lubumbashi", 0)

# 3. Affichage des résultats
print(f"Revenu moyen à Kinshasa : {moyenne_kinshasa:,.2f} USD")
print(f"Revenu moyen à Lubumbashi : {moyenne_lubumbashi:,.2f} USD\n")

# 4. Identification de la région la plus performante
if moyenne_kinshasa > moyenne_lubumbashi:
    ecart = moyenne_kinshasa - moyenne_lubumbashi
    print(f" Kinshasa performe mieux avec un revenu moyen supérieur de {ecart:,.2f} USD.")
elif moyenne_lubumbashi > moyenne_kinshasa:
    ecart = moyenne_lubumbashi - moyenne_kinshasa
    print(f" Lubumbashi performe mieux avec un revenu moyen supérieur de {ecart:,.2f} USD.")
else:
    print("Les deux régions ont exactement le même revenu moyen.")
```

 Calculer, pour chaque catégorie, la part du revenu total qu'elle représente à Kinshasa
uniquement

```python
# 1. Filtrer les données sur la région Kinshasa
df_kinshasa = df[df["Région"] == "Kinshasa"]

# 2. Calculer le revenu total à Kinshasa (toutes catégories confondues)
revenu_total_kinshasa = df_kinshasa["Revenu_USD"].sum()

# 3. Calculer la somme du revenu par catégorie
part_categories_kinshasa = (
    df_kinshasa.groupby("Catégorie")["Revenu_USD"]
    .sum()
    .reset_index()
)

# 4. Calculer le pourcentage et arrondir à 2 décimales
part_categories_kinshasa["Part_Revenu_Kinshasa (%)"] = (
    (part_categories_kinshasa["Revenu_USD"] / revenu_total_kinshasa) * 100
).round(2)

# 5. Trier par part décroissante
part_categories_kinshasa = part_categories_kinshasa.sort_values(
    by="Part_Revenu_Kinshasa (%)",
    ascending=False
)

# Affichage du résultat
part_categories_kinshasa
```

 Détecter s'il existe des valeurs manquantes ou aberrantes dans le dataset et proposer une
méthode pour les traiter

```python
# Nombre de valeurs manquantes par colonne
valeurs_manquantes = df.isnull().sum()

# Pourcentage de valeurs manquantes par colonne
pourcentage_manquant = (df.isnull().sum() / len(df)) * 100

# Tableau recapitulatif
df_nan = pd.DataFrame({
    'Manquants': valeurs_manquantes,
    'Pourcentage (%)': pourcentage_manquant.round(2)
})
df_nan[df_nan['Manquants'] > 0]
```

```python
# 1. Détection des valeurs négatives ou nulles (incohérences métiers)
incoherences = df[(df["Revenu_USD"] <= 0) | (df["Quantité"] <= 0)]
print(f"Nombre de ventes incohérentes (<= 0) : {len(incoherences)}")

# 2. Détection par la méthode de l'IQR (Écart interquartile) pour Revenu_USD
Q1 = df["Revenu_USD"].quantile(0.25)
Q3 = df["Revenu_USD"].quantile(0.75)
IQR = Q3 - Q1

# Bornes classique pour identifier les outliers
borne_inferieure = Q1 - 1.5 * IQR
borne_superieure = Q3 + 1.5 * IQR

outliers_iqr = df[(df["Revenu_USD"] < borne_inferieure) | (df["Revenu_USD"] > borne_superieure)]
print(f"Nombre d'outliers détectés par la méthode IQR : {len(outliers_iqr)}")
```
